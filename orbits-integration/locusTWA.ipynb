```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
""" Parameters """
N = 200
i0 = 15.5*np.pi/180
Rs = 3.5

epsilon = 1e-16
```

```python
phi_ = np.linspace(0, 360, 1000)
phi = phi_*np.pi/180

cota = np.log(2)/(N*np.tan(i0))
R1 = np.zeros(1000)

for i in range(len(phi)):
    if phi[i] >= cota:
        R1[i] = Rs*(np.exp(phi[i]*N*np.tan(i0)) - 1)**(1/N)

x_ = np.linspace(0, 10, 1000)
y_ = x_

x1 = R1*np.cos(phi)
y1 = R1*np.sin(phi)

x2 = -x1
y2 = -y1

plt.plot(x1, y1, x2, y2, lw = 0.8)
plt.plot(Rs*np.cos(np.linspace(0,2*np.pi, 1000)), Rs*np.sin(np.linspace(0,2*np.pi, 1000)) )
plt.axhline(y = 0, color='k', alpha = 0.6, lw = 0.8)
plt.axvline(x = 0, color='k', alpha = 0.6, lw = 0.8)
plt.axvline(x = Rs, color='r', alpha = 0.6, lw = 0.8)
xx = np.linspace(0,0.5, 100)
plt.plot(xx+Rs, xx/np.tan(i0), lw = 0.8)

plt.axis("equal")
plt.show()
```

```python
theta_ = np.array([1, 3, 5, 10, 15.5, 30, 45, 55])
theta = theta_*np.pi/180

R1 = np.zeros(1000)

for k in range(len(theta)):

    i0 = theta[k]
    
    phi_ = np.linspace(0, 360, 1000)
    phi = phi_*np.pi/180
    
    cota = np.log(2)/(N*np.tan(i0))
    
    for i in range(len(phi)):
        if phi[i] >= cota:
            R1[i] = Rs*(np.exp(phi[i]*N*np.tan(i0)) - 1)**(1/N)
    
    x_ = np.linspace(0, 10, 1000)
    y_ = x_
    
    x1 = R1*np.cos(phi)
    y1 = R1*np.sin(phi)
    
    x2 = -x1
    y2 = -y1
    
    plt.plot(x1, y1, x2, y2, lw = 0.8)
    plt.plot(Rs*np.cos(np.linspace(0,2*np.pi, 1000)), Rs*np.sin(np.linspace(0,2*np.pi, 1000)) )
    plt.axhline(y = 0, color='k', alpha = 0.6, lw = 0.8)
    plt.axvline(x = 0, color='k', alpha = 0.6, lw = 0.8)
    plt.axvline(x = Rs, color='r', alpha = 0.6, lw = 0.8)
    xx = np.linspace(0,0.5, 100)
    plt.plot(xx+Rs, xx/np.tan(i0), lw = 0.8)
    plt.title(f"Espiral con angulo de enrollamiento {theta_[k]}°")
    
    plt.axis("equal")
    plt.show()
```

```python
theta_ = 15.5
theta = theta_*np.pi/180

R1 = np.zeros(1000)

N = np.array([0 + epsilon, 1, 5, 10, 15, 20, 50, 100, 200, 300])

for k in range(len(N)):

    i0 = theta
    
    phi_ = np.linspace(0, 360, 1000)
    phi = phi_*np.pi/180
    
    cota = np.log(2)/(N[k]*np.tan(i0))
    
    for i in range(len(phi)):
        if phi[i] >= cota:
            R1[i] = Rs*(np.exp(phi[i]*N[k]*np.tan(i0)) - 1)**(1/N[k])
    
    x_ = np.linspace(0, 10, 1000)
    y_ = x_
    
    x1 = R1*np.cos(phi)
    y1 = R1*np.sin(phi)
    
    x2 = -x1
    y2 = -y1
    
    
    plt.plot(x1, y1, x2, y2, lw = 0.8)
    plt.plot(Rs*np.cos(np.linspace(0,2*np.pi, 1000)), Rs*np.sin(np.linspace(0,2*np.pi, 1000)) )
    plt.axhline(y = 0, color='k', alpha = 0.6, lw = 0.8)
    plt.axvline(x = 0, color='k', alpha = 0.6, lw = 0.8)
    plt.axvline(x = Rs, color='r', alpha = 0.6, lw = 0.8)
    xx = np.linspace(0,0.5, 100)
    plt.plot(xx+Rs, xx/np.tan(i0), lw = 0.8)
    if N[k] > epsilon:
        plt.title(f"Espiral con N = {N[k]}")
    else:
        plt.title(f"Espiral con N = 0.0")

    plt.axis("equal")
    plt.show()
```
